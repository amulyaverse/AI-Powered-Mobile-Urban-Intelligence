```python
from roboflow import Roboflow

# Download the dataset using your API key
rf = Roboflow(api_key="wr8L9ESEMlmrTRHJb8w7")
project = rf.workspace("project-o3ot9").project("indian-road-potholes")
version = project.version(5)
dataset = version.download("yolov8")

print(f"Dataset downloaded to: {dataset.location}")
```

```python
from ultralytics import YOLO
import os

# Load YOLOv8 Nano
model = YOLO("yolov8n.pt")

data_yaml = os.path.join(dataset.location, "data.yaml")

# Train with GPU settings
results = model.train(
    data=data_yaml,
    epochs=100,
    imgsz=640,
    batch=16,   # Increase to 32 or 64 if you have more VRAM (e.g. Kaggle P100/T4)
    device=0,   # Use GPU
    project="kaggle_pothole_training",
    name="pothole_model_cloud"
)
```
